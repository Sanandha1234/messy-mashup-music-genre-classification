```python
# ============================================
# CELL 1: IMPORTS AND SETUP (RUN ONCE)
# ============================================

import os
import random
import numpy as np
import pandas as pd
import librosa
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchaudio
import torchaudio.transforms as T
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from tqdm import tqdm
import glob
import warnings
warnings.filterwarnings('ignore')

# Wandb imports
!pip install wandb -q
import wandb
from kaggle_secrets import UserSecretsClient
import datetime

# Set seeds
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

# Device
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {DEVICE}")

# Constants
SR = 22050
DURATION = 30
MAX_LEN = SR * DURATION
N_MELS = 128
N_FFT = 2048
HOP_LENGTH = 512
BATCH_SIZE = 32
EPOCHS = 20  # For CRNN, you'll use 15 later

# Genres
GENRES = ["blues", "classical", "country", "disco", "hiphop",
          "jazz", "metal", "pop", "reggae", "rock"]
label2idx = {g: i for i, g in enumerate(GENRES)}
idx2label = {i: g for g, i in label2idx.items()}

# Paths
DATA_PATH = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup"
STEMS_PATH = os.path.join(DATA_PATH, "genres_stems")
NOISE_PATH = os.path.join(DATA_PATH, "ESC-50-master", "audio")
MASHUPS_PATH = os.path.join(DATA_PATH, "mashups")
TEST_CSV = os.path.join(DATA_PATH, "test.csv")

print(f"Data path: {DATA_PATH}")
print(f"Stems path: {STEMS_PATH}")
print(f"Noise path: {NOISE_PATH}")

# ============================================
# CELL 2: AUDIO PROCESSING FUNCTIONS (KEEP ONE COPY)
# ============================================

def load_audio(file_path, target_sr=SR):
    try:
        audio, sr = librosa.load(file_path, sr=target_sr, mono=True)
        return audio
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return np.zeros(target_sr * 30)

def estimate_tempo(audio, sr=SR):
    try:
        if isinstance(audio, torch.Tensor):
            audio = audio.numpy()
        tempo, _ = librosa.beat.beat_track(y=audio, sr=sr)
        if isinstance(tempo, np.ndarray):
            tempo = tempo[0] if len(tempo) > 0 else 120
        return float(tempo) if tempo > 0 else 120.0
    except:
        return 120.0

def apply_tempo_adjustment(audio, target_tempo, current_tempo, sr=SR):
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    target_tempo = float(target_tempo)
    current_tempo = float(current_tempo)
    
    if abs(current_tempo - target_tempo) < 5:
        return audio
    
    rate = target_tempo / current_tempo
    rate = np.clip(rate, 0.7, 1.3)
    
    try:
        stretched = librosa.effects.time_stretch(audio, rate=rate)
        return stretched
    except:
        return audio

def add_esc50_noise(audio, noise_files, num_noises=(1, 3), snr_range=(5, 20)):
    if len(noise_files) == 0:
        return audio
    
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    
    num_to_add = random.randint(*num_noises)
    noisy_audio = audio.copy()
    audio_len = len(audio)
    
    for _ in range(num_to_add):
        noise_path = random.choice(noise_files)
        noise, _ = librosa.load(noise_path, sr=SR)
        
        if len(noise) < audio_len:
            start_pos = random.randint(0, audio_len - len(noise))
            noise_padded = np.zeros(audio_len)
            noise_padded[start_pos:start_pos + len(noise)] = noise
        else:
            start_pos = random.randint(0, len(noise) - audio_len)
            noise_padded = noise[start_pos:start_pos + audio_len]
        
        snr_db = random.uniform(*snr_range)
        signal_power = np.mean(audio ** 2)
        noise_power = np.mean(noise_padded ** 2)
        
        if noise_power > 0:
            scale = np.sqrt(signal_power / (noise_power * (10 ** (snr_db / 10))))
            noisy_audio += scale * noise_padded
    
    max_val = np.max(np.abs(noisy_audio))
    if max_val > 1.0:
        noisy_audio = noisy_audio / max_val * 0.95
    
    return noisy_audio

def extract_melspectrogram(audio, sr=SR):
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    
    mel_spec = librosa.feature.melspectrogram(
        y=audio, sr=sr, n_mels=N_MELS,
        n_fft=N_FFT, hop_length=HOP_LENGTH
    )
    mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)
    mel_spec_db = (mel_spec_db - mel_spec_db.mean()) / (mel_spec_db.std() + 1e-8)
    return mel_spec_db


noise_files = glob.glob(os.path.join(NOISE_PATH, "*.wav"))
print(f"Found {len(noise_files)} noise files")

def build_samples(n_samples_per_genre=200):
    samples = []
    for genre in GENRES:
        genre_path = os.path.join(STEMS_PATH, genre)
        if not os.path.exists(genre_path):
            continue
        songs = os.listdir(genre_path)
        print(f"{genre}: {len(songs)} songs")
        
        for _ in range(n_samples_per_genre):
            if len(songs) < 4:
                selected = random.choices(songs, k=4)
            else:
                selected = random.sample(songs, 4)
            
            stems = []
            stem_names = ["drums.wav", "bass.wav", "vocals.wav", "other.wav"]
            for i, stem_name in enumerate(stem_names):
                stem_path = os.path.join(genre_path, selected[i], stem_name)
                stems.append(stem_path)
            
            samples.append({
                'stems': stems,
                'genre': genre,
                'label': label2idx[genre]
            })
    random.shuffle(samples)
    print(f"Total samples built: {len(samples)}")
    return samples

all_samples = build_samples(n_samples_per_genre=200)
train_samples, val_samples = train_test_split(
    all_samples, test_size=0.15, 
    stratify=[s['label'] for s in all_samples], random_state=42
)
print(f"\nTrain samples: {len(train_samples)}")
print(f"Validation samples: {len(val_samples)}")

class MashupDataset(Dataset):
    def __init__(self, samples, noise_files=None, is_train=True):
        self.samples = samples
        self.noise_files = noise_files or []
        self.is_train = is_train
        
        self.label_to_samples = {}
        for genre_idx in range(len(GENRES)):
            self.label_to_samples[genre_idx] = []
        for sample in samples:
            self.label_to_samples[sample['label']].append(sample['stems'])
    
    def __len__(self):
        return len(self.samples)
    
    def load_and_process_stem(self, stem_path):
        return load_audio(stem_path)
    
    def __getitem__(self, idx):
        sample = self.samples[idx]
        label = sample['label']
        
        if self.is_train:
            genre_samples = self.label_to_samples[label]
            if len(genre_samples) >= 4:
                selected = random.sample(genre_samples, 4) if len(genre_samples) >= 4 else random.choices(genre_samples, k=4)
                signals = []
                stem_names = ["drums.wav", "bass.wav", "vocals.wav", "other.wav"]
                for i, stem_name in enumerate(stem_names):
                    stem_folder = os.path.dirname(selected[i][0])
                    stem_path = os.path.join(stem_folder, stem_name)
                    sig = self.load_and_process_stem(stem_path)
                    gain = random.uniform(0.6, 1.5)
                    signals.append(sig * gain)
            else:
                signals = []
                for stem_path in sample['stems']:
                    sig = self.load_and_process_stem(stem_path)
                    gain = random.uniform(0.6, 1.5)
                    signals.append(sig * gain)
        else:
            signals = []
            for stem_path in sample['stems']:
                sig = self.load_and_process_stem(stem_path)
                signals.append(sig)
        
        signals = [np.array(sig) for sig in signals]
        min_len = min(len(sig) for sig in signals)
        signals = [sig[:min_len] for sig in signals]
        
        if self.is_train and random.random() > 0.5 and len(signals) > 1:
            try:
                tempo_drums = estimate_tempo(signals[0])
                for i in range(1, len(signals)):
                    tempo_stem = estimate_tempo(signals[i])
                    target_tempo = tempo_drums * random.uniform(0.9, 1.1)
                    adjusted = apply_tempo_adjustment(signals[i], target_tempo, tempo_stem)
                    if len(adjusted) > min_len:
                        signals[i] = adjusted[:min_len]
                    elif len(adjusted) < min_len:
                        signals[i] = np.pad(adjusted, (0, min_len - len(adjusted)))
                    else:
                        signals[i] = adjusted
            except:
                pass
        
        for i, sig in enumerate(signals):
            if len(sig) != min_len:
                if len(sig) > min_len:
                    signals[i] = sig[:min_len]
                else:
                    signals[i] = np.pad(sig, (0, min_len - len(sig)))
        
        mixed = np.sum(signals, axis=0)
        max_val = np.max(np.abs(mixed))
        if max_val > 0:
            mixed = mixed / max_val * 0.9
        
        if self.is_train and len(self.noise_files) > 0 and random.random() > 0.3:
            mixed = add_esc50_noise(mixed, self.noise_files, num_noises=(1, 2), snr_range=(8, 25))
        
        if len(mixed) > MAX_LEN:
            if self.is_train:
                start = random.randint(0, len(mixed) - MAX_LEN)
                mixed = mixed[start:start + MAX_LEN]
            else:
                start = (len(mixed) - MAX_LEN) // 2
                mixed = mixed[start:start + MAX_LEN]
        else:
            mixed = np.pad(mixed, (0, MAX_LEN - len(mixed)))
        
        features = extract_melspectrogram(mixed)
        features = torch.FloatTensor(features).unsqueeze(0)
        label = torch.LongTensor([label])[0]
        
        return features, label

# ============================================
# CELL 5: CREATE DATALOADERS
# ============================================

train_dataset = MashupDataset(train_samples, noise_files=noise_files, is_train=True)
val_dataset = MashupDataset(val_samples, noise_files=[], is_train=False)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

print(f"Train batches: {len(train_loader)}")
print(f"Val batches: {len(val_loader)}")

# ============================================
# CELL 6: WANDB LOGIN (RUN ONCE)
# ============================================

try:
    user_secrets = UserSecretsClient()
    wandb_key = user_secrets.get_secret("wandb_api_key")
    os.environ["WANDB_API_KEY"] = wandb_key
    print(" Wandb API key loaded from Kaggle secrets")
except Exception as e:
    print(f" Failed to load Wandb API key: {e}")
    YOUR_API_KEY = "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"
    os.environ["WANDB_API_KEY"] = YOUR_API_KEY
    print(" Using hardcoded API key")

# ============================================
# CELL 7: NORMAL CRNN MODEL (15 EPOCHS) WITH WANDB
# ============================================

class CRNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(), nn.MaxPool2d(2)
        )
        self.gru = nn.GRU(input_size=64*32, hidden_size=128, num_layers=1, batch_first=True, bidirectional=True)
        self.fc = nn.Sequential(nn.Dropout(0.5), nn.Linear(256, num_classes))

    def forward(self, x):
        x = self.cnn(x)
        b, c, f, t = x.size()
        x = x.permute(0, 3, 1, 2).reshape(b, t, c * f)
        x, _ = self.gru(x)
        x = x.mean(dim=1)
        return self.fc(x)

# Training functions
def train_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss = 0
    all_preds, all_labels = [], []
    
    for x, y in tqdm(loader, desc='Training'):
        x, y = x.to(DEVICE), y.to(DEVICE)
        optimizer.zero_grad()
        outputs = model(x)
        loss = criterion(outputs, y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
        optimizer.step()
        
        total_loss += loss.item()
        all_preds.extend(outputs.argmax(1).cpu().numpy())
        all_labels.extend(y.cpu().numpy())
    
    return total_loss/len(loader), f1_score(all_labels, all_preds, average='macro')

def validate(model, loader):
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            outputs = model(x)
            all_preds.extend(outputs.argmax(1).cpu().numpy())
            all_labels.extend(y.cpu().numpy())
    return f1_score(all_labels, all_preds, average='macro')

# Initialize Normal CRNN
print("\n" + "="*60)
print(" TRAINING NORMAL CRNN (15 EPOCHS)")
print("="*60)

model = CRNN(num_classes=10).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

wandb.init(
    project="DL-24f2008134-notebook-t12026",
    entity="sanandha0907-indian-institute-of-technology-madras",
    name=f"normal-crnn-15-epochs-{datetime.datetime.now().strftime('%Y%m%d-%H%M')}",
    tags=["crnn", "normal", "15-epochs"],
    config={
        "model_type": "Normal CRNN",
        "cnn_layers": 2,
        "gru_hidden": 128,
        "bidirectional": True,
        "batch_size": BATCH_SIZE,
        "learning_rate": 3e-4,
        "epochs": 15,
        "optimizer": "AdamW",
        "label_smoothing": 0.1
    },
    reinit=True
)

wandb.watch(model, log="all", log_freq=10)

best_f1 = 0
for epoch in range(15):
    train_loss, train_f1 = train_epoch(model, train_loader, optimizer, criterion)
    val_f1 = validate(model, val_loader)
    
    wandb.log({
        "epoch": epoch+1,
        "train_loss": train_loss,
        "train_f1": train_f1,
        "val_f1": val_f1,
        "learning_rate": optimizer.param_groups[0]['lr']
    })
    
    print(f"Epoch {epoch+1}/15 - Train F1: {train_f1:.4f}, Val F1: {val_f1:.4f}")
    
    if val_f1 > best_f1:
        best_f1 = val_f1
        torch.save(model.state_dict(), 'normal_crnn_best.pth')
        wandb.run.summary["best_val_f1"] = best_f1

wandb.finish()
print(f" Th best f1 for normal crnn architecture is :  Best F1: {best_f1:.4f}")

# ============================================
# CELL 8: IMPROVED CRNN MODEL (20 EPOCHS) WITH WANDB
# ============================================

class ImprovedCRNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        
        self.cnn = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(), nn.MaxPool2d(2), nn.Dropout2d(0.1),
            
            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(),
            nn.Conv2d(64, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(), nn.MaxPool2d(2), nn.Dropout2d(0.1),
            
            nn.Conv2d(64, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(),
            nn.Conv2d(128, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(), nn.MaxPool2d(2), nn.Dropout2d(0.1),
            
            nn.Conv2d(128, 256, 3, padding=1), nn.BatchNorm2d(256), nn.ReLU(),
            nn.AdaptiveAvgPool2d((None, 1))
        )
        
        self.rnn = nn.GRU(input_size=256, hidden_size=256, num_layers=2, batch_first=True, bidirectional=True, dropout=0.3)
        
        self.classifier = nn.Sequential(
            nn.LayerNorm(512), nn.Linear(512, 256), nn.ReLU(), nn.Dropout(0.4),
            nn.Linear(256, 128), nn.ReLU(), nn.Dropout(0.3), nn.Linear(128, num_classes)
        )
        
    def forward(self, x):
        x = self.cnn(x)
        x = x.squeeze(-1).transpose(1, 2)
        x, _ = self.rnn(x)
        x = x.mean(dim=1)
        return self.classifier(x)


print(" TRAINING IMPROVED CRNN (20 EPOCHS)")


model = ImprovedCRNN(num_classes=10).to(DEVICE)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
scaler = torch.amp.GradScaler('cuda')

wandb.init(
    project="DL-24f2008134-notebook-t12026",
    entity="sanandha0907-indian-institute-of-technology-madras",
    name=f"improved-crnn-20-epochs-{datetime.datetime.now().strftime('%Y%m%d-%H%M')}",
    tags=["crnn", "improved", "20-epochs"],
    config={
        "model_type": "Improved CRNN",
        "cnn_blocks": 4,
        "gru_layers": 2,
        "gru_hidden": 256,
        "bidirectional": True,
        "batch_size": BATCH_SIZE,
        "learning_rate": 3e-4,
        "epochs": 20,
        "optimizer": "AdamW",
        "label_smoothing": 0.1
    },
    reinit=True
)

wandb.watch(model, log="all", log_freq=10)

best_f1 = 0
patience = 5
patience_counter = 0

for epoch in range(20):
    train_loss, train_f1 = train_epoch(model, train_loader, optimizer, criterion)
    val_f1 = validate(model, val_loader)
    
    wandb.log({
        "epoch": epoch+1,
        "train_loss": train_loss,
        "train_f1": train_f1,
        "val_f1": val_f1,
        "learning_rate": optimizer.param_groups[0]['lr']
    })
    
    print(f"Epoch {epoch+1}/20 - Train F1: {train_f1:.4f}, Val F1: {val_f1:.4f}")
    
    if val_f1 > best_f1:
        best_f1 = val_f1
        patience_counter = 0
        torch.save(model.state_dict(), 'improved_crnn_best.pth')
        wandb.run.summary["best_val_f1"] = best_f1
        print(f" Saved best model (F1: {best_f1:.4f})")
    else:
        patience_counter += 1
    
    if patience_counter >= patience:
        print(f"Early stopping at epoch {epoch+1}")
        break

wandb.finish()
print(f"  Best F1: {best_f1:.4f}")

# ============================================
# CELL 9: INFERENCE AND SUBMISSION (OPTIONAL)
# ============================================

print("\n" + "="*60)
print(" The submission file is ")
print("="*60)

test_df = pd.read_csv(TEST_CSV)
print(f"Test samples: {len(test_df)}")

# Use the best model (Improved CRNN)
inference_model = ImprovedCRNN(num_classes=10).to(DEVICE)
inference_model.load_state_dict(torch.load('improved_crnn_best.pth'))
inference_model.eval()

predictions = []
for idx, row in tqdm(test_df.iterrows(), total=len(test_df)):
    path = os.path.join(DATA_PATH, row['filename'])
    
    audio, _ = librosa.load(path, sr=SR)
    if len(audio) > MAX_LEN:
        start = (len(audio) - MAX_LEN) // 2
        audio = audio[start:start + MAX_LEN]
    else:
        audio = np.pad(audio, (0, MAX_LEN - len(audio)))
    
    features = extract_melspectrogram(audio)
    features = torch.FloatTensor(features).unsqueeze(0).unsqueeze(0).to(DEVICE)
    
    with torch.no_grad():
        output = inference_model(features)
        pred = output.argmax(1).item()
    
    predictions.append(idx2label[pred])

submission = pd.DataFrame({'id': test_df['id'], 'genre': predictions})
submission.to_csv('submission.csv', index=False)
print(f" The file submission is: Shape: {submission.shape}")
print(submission.head(10))
```
