```python
DATA_PATH = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup"
STEMS_PATH = os.path.join(DATA_PATH, "genres_stems")
NOISE_PATH = os.path.join(DATA_PATH, "ESC-50-master/audio")
MASHUP_PATH = os.path.join(DATA_PATH, "mashups")

GENRES = ["blues","classical","country","disco","hiphop",
          "jazz","metal","pop","reggae","rock"]

label2idx = {g:i for i,g in enumerate(GENRES)}
idx2label = {i:g for g,i in label2idx.items()}
```

```python
def build_training_samples(n_samples=2000):
    samples = []

    for genre in GENRES:
        genre_path = os.path.join(STEMS_PATH, genre)
        songs = os.listdir(genre_path)

        for _ in range(n_samples // len(GENRES)):
            selected = random.sample(songs, 4)

            stems = []
            for i, stem_name in enumerate(["drums.wav","bass.wav","vocals.wav","other.wav"]):
                song_folder = os.path.join(genre_path, selected[i])
                stems.append(os.path.join(song_folder, stem_name))

            samples.append((stems, label2idx[genre]))

    random.shuffle(samples)
    return samples

samples = build_training_samples(2000)
```

```python
train_samples, val_samples = train_test_split(
    samples, test_size=0.2, stratify=[s[1] for s in samples], random_state=42
)
```

```python
class MashupDataset(torch.utils.data.Dataset):
    def __init__(self, samples, duration=30, sr=22050):
        self.samples = samples
        self.sr = sr
        self.duration = duration

        self.mel = T.MelSpectrogram(
            sample_rate=sr,
            n_fft=2048,
            hop_length=512,
            n_mels=128
        )

    def load_audio(self, path):
        audio, sr = torchaudio.load(path)
        audio = torchaudio.functional.resample(audio, sr, self.sr)
        audio = audio.mean(0)
        return audio

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        stems, label = self.samples[idx]

        signals = []
        for s in stems:
            sig = self.load_audio(s)
            signals.append(sig)

        min_len = min([len(s) for s in signals])
        signals = [s[:min_len] for s in signals]

        mixed = sum(signals)

        # Add mild noise
        if random.random() < 0.3:
            noise = 0.003 * torch.randn_like(mixed)
            mixed = mixed + noise

        # Trim/pad
        max_len = self.sr * self.duration
        if len(mixed) > max_len:
            mixed = mixed[:max_len]
        else:
            pad = max_len - len(mixed)
            mixed = torch.nn.functional.pad(mixed, (0,pad))

        mel = self.mel(mixed)
        mel = torch.log(mel + 1e-6)
        mel = (mel - mel.mean()) / (mel.std() + 1e-6)

        return mel.unsqueeze(0), label
```

```python
BATCH_SIZE = 16

train_loader = torch.utils.data.DataLoader(
    MashupDataset(train_samples),
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = torch.utils.data.DataLoader(
    MashupDataset(val_samples),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
```

```python
class CRNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.gru = nn.GRU(
            input_size=64*32,
            hidden_size=128,
            num_layers=1,
            batch_first=True,
            bidirectional=True
        )

        self.fc = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.cnn(x)
        b,c,f,t = x.size()
        x = x.permute(0,3,1,2)
        x = x.reshape(b,t,c*f)
        x,_ = self.gru(x)
        x = x.mean(dim=1)
        return self.fc(x)

model = CRNN().to(DEVICE)
```

```python
optimizer = optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=20)
```

```python
def train_epoch(loader):
    model.train()
    total_loss = 0
    preds, targets = [], []

    for x,y in tqdm(loader):
        x,y = x.to(DEVICE), y.to(DEVICE)

        optimizer.zero_grad()
        out = model(x)
        loss = criterion(out,y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        preds.extend(out.argmax(1).cpu().numpy())
        targets.extend(y.cpu().numpy())

    f1 = f1_score(targets, preds, average="macro")
    return total_loss/len(loader), f1


def validate(loader):
    model.eval()
    preds, targets = [], []

    with torch.no_grad():
        for x,y in loader:
            x,y = x.to(DEVICE), y.to(DEVICE)
            out = model(x)
            preds.extend(out.argmax(1).cpu().numpy())
            targets.extend(y.cpu().numpy())

    f1 = f1_score(targets, preds, average="macro")
    return f1
```

```python
EPOCHS = 3
best_f1 = 0

for epoch in range(EPOCHS):
    train_loss, train_f1 = train_epoch(train_loader)
    val_f1 = validate(val_loader)
    scheduler.step()

    print(f"Epoch {epoch+1}")
    print("Train F1:", train_f1)
    print("Val F1:", val_f1)

    if val_f1 > best_f1:
        best_f1 = val_f1
        torch.save(model.state_dict(), "best_model.pth")

print("Best Val F1:", best_f1)
```

```python
model.load_state_dict(torch.load("best_model.pth"))
model.eval()

test_df = pd.read_csv(os.path.join(DATA_PATH, "test.csv"))

mel_transform = T.MelSpectrogram(
    sample_rate=22050,
    n_fft=2048,
    hop_length=512,
    n_mels=128
)

predictions = []

for file_path in tqdm(test_df["filename"]):
    
    # Full correct path
    path = os.path.join(DATA_PATH, file_path)

    audio, sr = torchaudio.load(path)
    audio = torchaudio.functional.resample(audio, sr, 22050)
    audio = audio.mean(0)

    mel = mel_transform(audio)
    mel = torch.log(mel + 1e-6)
    mel = (mel - mel.mean()) / (mel.std() + 1e-6)

    mel = mel.unsqueeze(0).unsqueeze(0).to(DEVICE)

    with torch.no_grad():
        out = model(mel)
        pred = out.argmax(1).item()

    predictions.append(idx2label[pred])
```
