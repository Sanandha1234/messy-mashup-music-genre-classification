```python
#Inference test submission

print(" LOADING MODELS AND CREATING ENSEMBLE SUBMISSION")

import os
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import librosa
import torchaudio
from transformers import ASTForAudioClassification, WavLMForSequenceClassification, AutoFeatureExtractor
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {DEVICE}")

SR = 22050
SR_TRANSFORMER = 16000
DURATION = 30
MAX_LEN = SR * DURATION
NUM_CLASSES = 10

GENRES = ["blues", "classical", "country", "disco", "hiphop",
          "jazz", "metal", "pop", "reggae", "rock"]
idx2label = {i: g for i, g in enumerate(GENRES)}

DATA_PATH = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup"
TEST_CSV = os.path.join(DATA_PATH, "test.csv")

test_df = pd.read_csv(TEST_CSV)
print(f"Test samples: {len(test_df)}")

feature_extractor = AutoFeatureExtractor.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593")

class WavLMClassifier(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.wavlm = WavLMForSequenceClassification.from_pretrained(
            "microsoft/wavlm-base-plus",
            num_labels=num_classes,
            ignore_mismatched_sizes=True
        )
    def forward(self, x):
        return self.wavlm(x).logits

def load_ast_model(model_path):
    model = ASTForAudioClassification.from_pretrained(
        "MIT/ast-finetuned-audioset-10-10-0.4593",
        num_labels=NUM_CLASSES,
        ignore_mismatched_sizes=True
    ).to(DEVICE)
    model.load_state_dict(torch.load(model_path, map_location=DEVICE))
    model.eval()
    print("AST model loaded")
    return model

def load_wavlm_model(model_path):
    model = WavLMClassifier(num_classes=10).to(DEVICE)
    model.load_state_dict(torch.load(model_path, map_location=DEVICE))
    model.eval()
    print("WavLM model loaded")
    return model

def process_audio_for_ast(audio_path):
    audio, _ = librosa.load(audio_path, sr=SR)
    if len(audio) > MAX_LEN:
        start = (len(audio) - MAX_LEN) // 2
        audio = audio[start:start + MAX_LEN]
    else:
        audio = np.pad(audio, (0, MAX_LEN - len(audio)))
    audio_tensor = torch.tensor(audio).float()
    audio_tensor = torchaudio.functional.resample(audio_tensor, SR, SR_TRANSFORMER)
    audio_np = audio_tensor.numpy()
    inputs = feature_extractor(
        audio_np,
        sampling_rate=SR_TRANSFORMER,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=SR_TRANSFORMER * DURATION
    )
    return inputs["input_values"]

def process_audio_for_wavlm(audio_path):
    audio, _ = librosa.load(audio_path, sr=SR)
    if len(audio) > MAX_LEN:
        start = (len(audio) - MAX_LEN) // 2
        audio = audio[start:start + MAX_LEN]
    else:
        audio = np.pad(audio, (0, MAX_LEN - len(audio)))
    audio_tensor = torch.tensor(audio).float()
    audio_tensor = torchaudio.functional.resample(audio_tensor, SR, SR_TRANSFORMER)
    if audio_tensor.abs().max() > 0:
        audio_tensor = audio_tensor / (audio_tensor.abs().max() + 1e-6)
    return audio_tensor.unsqueeze(0)

def predict_ast(model, test_df):
    probs = []
    with torch.no_grad():
        for idx, row in tqdm(test_df.iterrows(), total=len(test_df), desc="AST Inference"):
            audio_path = os.path.join(DATA_PATH, row['filename'])
            input_values = process_audio_for_ast(audio_path).to(DEVICE)
            outputs = model(input_values).logits
            probs.append(torch.softmax(outputs, dim=1).cpu().numpy()[0])
    return np.array(probs)

def predict_wavlm(model, test_df):
    probs = []
    with torch.no_grad():
        for idx, row in tqdm(test_df.iterrows(), total=len(test_df), desc="WavLM Inference"):
            audio_path = os.path.join(DATA_PATH, row['filename'])
            x = process_audio_for_wavlm(audio_path).to(DEVICE)
            outputs = model(x)
            probs.append(torch.softmax(outputs, dim=1).cpu().numpy()[0])
    return np.array(probs)
print(" LOADING MODELS")
ast_model = load_ast_model('/kaggle/input/datasets/psanandha/ast-wavlm/ast_best.pth')
wavlm_model = load_wavlm_model('/kaggle/input/datasets/psanandha/wavlm-new/wavlm_optimized.pth')
print(" GENERATING PREDICTIONS")
ast_probs = predict_ast(ast_model, test_df)
np.save('ast_probs.npy', ast_probs)
print(f"AST predictions saved: {ast_probs.shape}")
wavlm_probs = predict_wavlm(wavlm_model, test_df)
np.save('wavlm_probs.npy', wavlm_probs)
print(f"WavLM predictions saved: {wavlm_probs.shape}")
AST_WEIGHT = 0.70
WAVLM_WEIGHT = 0.30
print(f"Ensemble weights:")
print(f"  AST: {AST_WEIGHT:.2f}")
print(f"  WavLM: {WAVLM_WEIGHT:.2f}")

ensemble_probs = AST_WEIGHT * ast_probs + WAVLM_WEIGHT * wavlm_probs
final_preds = ensemble_probs.argmax(1)
final_genres = [idx2label[p] for p in final_preds]

submission = pd.DataFrame({'id': test_df['id'], 'genre': final_genres})
submission.to_csv('submission_AST70_wavLM30.csv', index=False)

print(f"\nSubmission saved as 'submission_AST70_wavLM30.csv'")
print(f"Shape: {submission.shape}")
print("\nFirst 10 predictions:")
print(submission.head(10))

print("\nGenre distribution:")
print(submission['genre'].value_counts().sort_index())
```
