```python
DATA_PATH = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup"
STEMS_PATH = os.path.join(DATA_PATH, "genres_stems")
NOISE_PATH = os.path.join(DATA_PATH, "ESC-50-master/audio")
MASHUP_PATH = os.path.join(DATA_PATH, "mashups")

GENRES = ["blues","classical","country","disco","hiphop",
          "jazz","metal","pop","reggae","rock"]

label2idx = {g:i for i,g in enumerate(GENRES)}
idx2label = {i:g for g,i in label2idx.items()}
```

```python
def build_training_samples(n_samples=2000):
    samples = []

    for genre in GENRES:
        genre_path = os.path.join(STEMS_PATH, genre)
        songs = os.listdir(genre_path)

        for _ in range(n_samples // len(GENRES)):
            selected = random.sample(songs, 4)

            stems = []
            for i, stem_name in enumerate(["drums.wav","bass.wav","vocals.wav","other.wav"]):
                song_folder = os.path.join(genre_path, selected[i])
                stems.append(os.path.join(song_folder, stem_name))

            samples.append((stems, label2idx[genre]))

    random.shuffle(samples)
    return samples

samples = build_training_samples(2000)
```

```python
train_samples, val_samples = train_test_split(
    samples, test_size=0.2, stratify=[s[1] for s in samples], random_state=42
)
```

```python
import random
import torch
import torchaudio
import torchaudio.transforms as T
import torch.nn.functional as F
from collections import defaultdict

class MashupDataset(torch.utils.data.Dataset):
    def __init__(self, samples, duration=30, sr=22050, train=True):
        self.samples = samples
        self.sr = sr
        self.duration = duration
        self.train = train

        # Group by label for cross-song mixing
        self.label_to_samples = defaultdict(list)
        for stems, label in samples:
            self.label_to_samples[label].append(stems)

        self.mel = T.MelSpectrogram(
            sample_rate=sr,
            n_fft=2048,
            hop_length=512,
            n_mels=128
        )

        self.freq_mask = T.FrequencyMasking(15)
        self.time_mask = T.TimeMasking(35)

    def load_audio(self, path):
        audio, sr = torchaudio.load(path)
        audio = torchaudio.functional.resample(audio, sr, self.sr)
        audio = audio.mean(0)
        return audio

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        stems, label = self.samples[idx]
        signals = []

        if self.train:
            # 🔥 Cross-song mixing
            candidates = self.label_to_samples[label]
            chosen = random.choices(candidates, k=len(stems))

            for i in range(len(stems)):
                sig = self.load_audio(chosen[i][i])
                gain = random.uniform(0.6, 1.4)
                sig *= gain
                signals.append(sig)
        else:
            for s in stems:
                signals.append(self.load_audio(s))

        min_len = min(len(s) for s in signals)
        signals = [s[:min_len] for s in signals]
        mixed = sum(signals)

        # Normalize
        mixed = mixed / (mixed.abs().max() + 1e-6)

        # Noise
        if self.train and random.random() < 0.6:
            noise = torch.randn_like(mixed)
            noise_level = random.uniform(0.01, 0.03)
            mixed += noise * noise_level

        max_len = self.sr * self.duration

        if len(mixed) > max_len:
            if self.train:
                start = random.randint(0, len(mixed) - max_len)
            else:
                start = 0
            mixed = mixed[start:start+max_len]
        else:
            mixed = F.pad(mixed, (0, max_len - len(mixed)))

        mel = self.mel(mixed)
        mel = torch.log(mel + 1e-6)
        mel = (mel - mel.mean()) / (mel.std() + 1e-6)

        if self.train:
            mel = self.freq_mask(mel)
            mel = self.time_mask(mel)

        return mel.unsqueeze(0), label
```

```python
BATCH_SIZE = 16

train_loader = torch.utils.data.DataLoader(
    MashupDataset(train_samples),
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = torch.utils.data.DataLoader(
    MashupDataset(val_samples),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)
```

```python
class CRNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()

        self.cnn = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.gru = nn.GRU(
            input_size=64*32,
            hidden_size=128,
            num_layers=1,
            batch_first=True,
            bidirectional=True
        )

        self.fc = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.cnn(x)
        b,c,f,t = x.size()
        x = x.permute(0,3,1,2)
        x = x.reshape(b,t,c*f)
        x,_ = self.gru(x)
        x = x.mean(dim=1)
        return self.fc(x)

model = CRNN().to(DEVICE)
```

```python
import torch.optim as optim
import torch.nn as nn

EPOCHS = 20

optimizer = optim.AdamW(
    model.parameters(),
    lr=3e-4,
    weight_decay=1e-4
)

criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=EPOCHS
)
```

```python
from torch.cuda.amp import autocast, GradScaler

scaler = GradScaler()

def train_epoch(loader):
    model.train()
    total_loss = 0
    preds, targets = [], []

    for x, y in tqdm(loader):
        x, y = x.to(DEVICE), y.to(DEVICE)

        optimizer.zero_grad()

        with autocast():  # mixed precision
            out = model(x)
            loss = criterion(out, y)

        scaler.scale(loss).backward()

        # Gradient clipping
        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)

        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item()
        preds.extend(out.argmax(1).detach().cpu().numpy())
        targets.extend(y.cpu().numpy())

    f1 = f1_score(targets, preds, average="macro")
    return total_loss / len(loader), f1
```

```python
def validate(loader):
    model.eval()
    total_loss = 0
    preds, targets = [], []

    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            out = model(x)
            loss = criterion(out, y)

            total_loss += loss.item()
            preds.extend(out.argmax(1).cpu().numpy())
            targets.extend(y.cpu().numpy())

    f1 = f1_score(targets, preds, average="macro")
    return f1
```

```python
best_f1 = 0.0
patience = 5
patience_counter = 0

for epoch in range(EPOCHS):

    train_loss, train_f1 = train_epoch(train_loader)
    val_f1 = validate(val_loader)

    print(f"\nEpoch {epoch+1}")
    print("Train F1:", train_f1)
    print("Val F1:", val_f1)
    print("LR:", optimizer.param_groups[0]['lr'])

    if val_f1 > best_f1:
        best_f1 = val_f1
        patience_counter = 0
        torch.save(model.state_dict(), "/kaggle/working/best_model.pth")
        print("Model Saved!")
    else:
        patience_counter += 1

    scheduler.step()

    if patience_counter >= patience:
        print("Early stopping triggered.")
        break

print("\nBest Val F1:", best_f1)
```

```python
# ============================================
# CELL 2: COMPLETE AUDIO PROCESSING FUNCTIONS
# ============================================

def load_audio(file_path, target_sr=SR):
    """Load audio file with proper resampling and mono conversion"""
    try:
        audio, sr = librosa.load(file_path, sr=target_sr, mono=True)
        return audio
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return np.zeros(target_sr * 30)

def estimate_tempo(audio, sr=SR):
    """Estimate tempo of audio using librosa"""
    try:
        # Ensure audio is numpy array
        if isinstance(audio, torch.Tensor):
            audio = audio.numpy()
        
        tempo, _ = librosa.beat.beat_track(y=audio, sr=sr)
        # Handle case where tempo is array
        if isinstance(tempo, np.ndarray):
            tempo = tempo[0] if len(tempo) > 0 else 120
        return float(tempo) if tempo > 0 else 120.0
    except Exception as e:
        print(f"Tempo estimation error: {e}")
        return 120.0

def apply_tempo_adjustment(audio, target_tempo, current_tempo, sr=SR):
    """Apply tempo adjustment to match target tempo"""
    # Ensure audio is numpy array
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    
    # Convert to float
    target_tempo = float(target_tempo)
    current_tempo = float(current_tempo)
    
    if abs(current_tempo - target_tempo) < 5:  # Less than 5 BPM difference
        return audio
    
    rate = target_tempo / current_tempo
    rate = np.clip(rate, 0.7, 1.3)  # Limit to reasonable range
    
    try:
        # Apply time stretch
        stretched = librosa.effects.time_stretch(audio, rate=rate)
        return stretched
    except Exception as e:
        print(f"Tempo adjustment error: {e}")
        return audio

def add_esc50_noise(audio, noise_files, num_noises=(1, 3), snr_range=(5, 20)):
    """Add real ESC-50 noise exactly like test data"""
    if len(noise_files) == 0:
        return audio
    
    # Ensure audio is numpy array
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    
    num_to_add = random.randint(*num_noises)
    noisy_audio = audio.copy()
    audio_len = len(audio)
    
    for _ in range(num_to_add):
        # Load random noise
        noise_path = random.choice(noise_files)
        noise, _ = librosa.load(noise_path, sr=SR)
        
        # Random position
        if len(noise) < audio_len:
            start_pos = random.randint(0, audio_len - len(noise))
            noise_padded = np.zeros(audio_len)
            noise_padded[start_pos:start_pos + len(noise)] = noise
        else:
            start_pos = random.randint(0, len(noise) - audio_len)
            noise_padded = noise[start_pos:start_pos + audio_len]
        
        # Random SNR
        snr_db = random.uniform(*snr_range)
        signal_power = np.mean(audio ** 2)
        noise_power = np.mean(noise_padded ** 2)
        
        if noise_power > 0:
            scale = np.sqrt(signal_power / (noise_power * (10 ** (snr_db / 10))))
            noisy_audio += scale * noise_padded
    
    # Normalize to prevent clipping
    max_val = np.max(np.abs(noisy_audio))
    if max_val > 1.0:
        noisy_audio = noisy_audio / max_val * 0.95
    
    return noisy_audio

def extract_melspectrogram(audio, sr=SR):
    """Extract mel-spectrogram features"""
    # Ensure audio is numpy array
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    
    # Compute mel spectrogram
    mel_spec = librosa.feature.melspectrogram(
        y=audio, 
        sr=sr, 
        n_mels=N_MELS,
        n_fft=N_FFT, 
        hop_length=HOP_LENGTH
    )
    
    # Convert to dB scale
    mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)
    
    # Normalize
    mel_spec_db = (mel_spec_db - mel_spec_db.mean()) / (mel_spec_db.std() + 1e-8)
    
    return mel_spec_db
```

```python
# ============================================
# CELL 4: DATASET CLASS WITH PROPER AUGMENTATION (FIXED)
# ============================================

class MashupDataset(Dataset):
    def __init__(self, samples, noise_files=None, is_train=True):
        self.samples = samples
        self.noise_files = noise_files or []
        self.is_train = is_train
        
        # Group by label for cross-song mixing during training
        self.label_to_samples = {}
        for genre_idx in range(len(GENRES)):
            self.label_to_samples[genre_idx] = []
        
        for sample in samples:
            self.label_to_samples[sample['label']].append(sample['stems'])
    
    def __len__(self):
        return len(self.samples)
    
    def load_and_process_stem(self, stem_path):
        """Load a single stem"""
        audio = load_audio(stem_path)
        return audio
    
    def __getitem__(self, idx):
        sample = self.samples[idx]
        label = sample['label']
        
        if self.is_train:
            # TRAINING: Cross-song mixing with augmentation
            # Get stems from different songs of same genre
            genre_samples = self.label_to_samples[label]
            
            if len(genre_samples) >= 4:
                # Select 4 different song folders
                selected = random.sample(genre_samples, 4) if len(genre_samples) >= 4 else random.choices(genre_samples, k=4)
                
                # Load stems from different songs
                signals = []
                stem_names = ["drums.wav", "bass.wav", "vocals.wav", "other.wav"]
                
                for i, stem_name in enumerate(stem_names):
                    # Get the path for this stem from the selected song
                    stem_folder = os.path.dirname(selected[i][0])
                    stem_path = os.path.join(stem_folder, stem_name)
                    sig = self.load_and_process_stem(stem_path)
                    
                    # Random gain (instrument balance)
                    gain = random.uniform(0.6, 1.5)
                    sig = sig * gain
                    
                    signals.append(sig)
            else:
                # Fallback to original stems
                signals = []
                for stem_path in sample['stems']:
                    sig = self.load_and_process_stem(stem_path)
                    gain = random.uniform(0.6, 1.5)
                    signals.append(sig * gain)
        else:
            # VALIDATION: Simple mix without augmentation
            signals = []
            for stem_path in sample['stems']:
                sig = self.load_and_process_stem(stem_path)
                signals.append(sig)
        
        # FIX: Ensure all signals are numpy arrays and find target length
        signals = [np.array(sig) for sig in signals]
        
        # Find a common length (use minimum of all signals)
        min_len = min(len(sig) for sig in signals)
        
        # Trim all signals to the same length
        signals = [sig[:min_len] for sig in signals]
        
        # Estimate tempo for drums (first stem) - only if training and random chance
        if self.is_train and random.random() > 0.5 and len(signals) > 1:
            try:
                tempo_drums = estimate_tempo(signals[0])
                
                # Apply tempo adjustments to other stems
                for i in range(1, len(signals)):
                    tempo_stem = estimate_tempo(signals[i])
                    # Slight random variation
                    target_tempo = tempo_drums * random.uniform(0.9, 1.1)
                    adjusted = apply_tempo_adjustment(signals[i], target_tempo, tempo_stem)
                    
                    # FIX: After tempo adjustment, ensure length matches
                    if len(adjusted) > min_len:
                        signals[i] = adjusted[:min_len]
                    elif len(adjusted) < min_len:
                        signals[i] = np.pad(adjusted, (0, min_len - len(adjusted)))
                    else:
                        signals[i] = adjusted
            except Exception as e:
                # If tempo adjustment fails, keep original
                pass
        
        # FIX: Double-check all signals have the same length
        for i, sig in enumerate(signals):
            if len(sig) != min_len:
                if len(sig) > min_len:
                    signals[i] = sig[:min_len]
                else:
                    signals[i] = np.pad(sig, (0, min_len - len(sig)))
        
        # Mix stems (now all same length)
        mixed = np.sum(signals, axis=0)
        
        # Normalize
        max_val = np.max(np.abs(mixed))
        if max_val > 0:
            mixed = mixed / max_val * 0.9
        
        # Add ESC-50 noise during training
        if self.is_train and len(self.noise_files) > 0 and random.random() > 0.3:
            mixed = add_esc50_noise(mixed, self.noise_files, 
                                    num_noises=(1, 2), snr_range=(8, 25))
        
        # Pad/trim to fixed length
        if len(mixed) > MAX_LEN:
            if self.is_train:
                # Random crop for training
                start = random.randint(0, len(mixed) - MAX_LEN)
                mixed = mixed[start:start + MAX_LEN]
            else:
                # Center crop for validation
                start = (len(mixed) - MAX_LEN) // 2
                mixed = mixed[start:start + MAX_LEN]
        else:
            mixed = np.pad(mixed, (0, MAX_LEN - len(mixed)))
        
        # Extract features
        features = extract_melspectrogram(mixed)
        
        # Convert to tensors
        features = torch.FloatTensor(features).unsqueeze(0)
        label = torch.LongTensor([label])[0]
        
        return features, label
```

```python
# ============================================
# CELL 5: CREATE TEST-LIKE VALIDATION SET (FIXED)
# ============================================

def create_test_like_validation(n_samples_per_genre=20):
    """Create validation set that EXACTLY matches test distribution"""
    val_samples = []
    
    for genre in GENRES:
        genre_path = os.path.join(STEMS_PATH, genre)
        if not os.path.exists(genre_path):
            continue
            
        all_songs = os.listdir(genre_path)
        
        # Use songs not heavily used in training
        val_songs = all_songs[-30:] if len(all_songs) > 30 else all_songs
        
        for _ in range(n_samples_per_genre):
            try:
                # Select stems from different songs
                if len(val_songs) < 4:
                    selected = random.choices(val_songs, k=4)
                else:
                    selected = random.sample(val_songs, 4)
                
                # Load all stems
                signals = []
                stem_names = ["drums.wav", "bass.wav", "vocals.wav", "other.wav"]
                
                for i, stem_name in enumerate(stem_names):
                    stem_path = os.path.join(genre_path, selected[i], stem_name)
                    audio = load_audio(stem_path)
                    signals.append(audio)
                
                # Align lengths
                min_len = min(len(s) for s in signals)
                signals = [s[:min_len] for s in signals]
                
                # Apply tempo adjustments to match drums
                tempo_drums = estimate_tempo(signals[0])
                for i in range(1, len(signals)):
                    tempo_stem = estimate_tempo(signals[i])
                    signals[i] = apply_tempo_adjustment(signals[i], tempo_drums, tempo_stem)
                
                # Mix with random gains
                gains = [random.uniform(0.7, 1.3) for _ in range(4)]
                mixed = np.zeros_like(signals[0])
                for i, signal in enumerate(signals):
                    # Ensure same length
                    min_len_mix = min(len(mixed), len(signal))
                    mixed[:min_len_mix] += signal[:min_len_mix] * gains[i]
                
                # Normalize
                max_val = np.max(np.abs(mixed))
                if max_val > 0:
                    mixed = mixed / max_val * 0.9
                
                # Add ESC-50 noise (1-3 noises)
                if len(noise_files) > 0:
                    mixed = add_esc50_noise(mixed, noise_files, num_noises=(1, 3), snr_range=(5, 20))
                
                # Pad/trim to fixed length
                if len(mixed) > MAX_LEN:
                    start = (len(mixed) - MAX_LEN) // 2
                    mixed = mixed[start:start + MAX_LEN]
                else:
                    mixed = np.pad(mixed, (0, MAX_LEN - len(mixed)))
                
                # Extract features
                features = extract_melspectrogram(mixed)
                
                val_samples.append({
                    'features': features,
                    'label': label2idx[genre]
                })
            except Exception as e:
                print(f"Error creating validation sample: {e}")
                continue
    
    print(f"Created {len(val_samples)} test-like validation samples")
    return val_samples

# Create test-like validation set
print("\nCreating test-like validation set...")
test_like_val_samples = create_test_like_validation(n_samples_per_genre=20)
```

```python
# ============================================
# CELL 6: IMPROVED MODEL ARCHITECTURE
# ============================================

class ImprovedCRNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        
        # CNN feature extractor
        self.cnn = nn.Sequential(
            # Block 1
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.1),
            
            # Block 2
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.1),
            
            # Block 3
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.Conv2d(128, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout2d(0.1),
            
            # Block 4
            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((None, 1))  # Pool frequency dimension
        )
        
        # RNN for temporal modeling
        self.rnn = nn.GRU(
            input_size=256,
            hidden_size=256,
            num_layers=2,
            batch_first=True,
            bidirectional=True,
            dropout=0.3
        )
        
        # Classifier
        self.classifier = nn.Sequential(
            nn.LayerNorm(512),  # 256*2 for bidirectional
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, num_classes)
        )
        
    def forward(self, x):
        # CNN
        x = self.cnn(x)  # (batch, channels, time, 1)
        x = x.squeeze(-1).transpose(1, 2)  # (batch, time, channels)
        
        # RNN
        x, _ = self.rnn(x)  # (batch, time, 512)
        
        # Global average pooling over time
        x = x.mean(dim=1)  # (batch, 512)
        
        # Classifier
        x = self.classifier(x)
        
        return x
```

```python
# ============================================
# CELL 7: DATALOADERS
# ============================================

# Create datasets
print("\nCreating datasets...")
train_dataset = MashupDataset(train_samples, noise_files=noise_files, is_train=True)
val_dataset = MashupDataset(val_samples, noise_files=[], is_train=False)

# Test-like validation dataset (pre-computed)
class PrecomputedDataset(Dataset):
    def __init__(self, samples):
        self.samples = samples
    
    def __len__(self):
        return len(self.samples)
    
    def __getitem__(self, idx):
        sample = self.samples[idx]
        features = torch.FloatTensor(sample['features']).unsqueeze(0)
        label = torch.LongTensor([sample['label']])[0]
        return features, label

if len(test_like_val_samples) > 0:
    test_like_val_dataset = PrecomputedDataset(test_like_val_samples)
else:
    print("Warning: No test-like validation samples created, using regular validation")
    test_like_val_dataset = val_dataset

# Create dataloaders
BATCH_SIZE = 32
NUM_WORKERS = 2

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_like_val_loader = DataLoader(
    test_like_val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

print("Dataloaders created successfully")
print(f"Train batches: {len(train_loader)}")
print(f"Val batches: {len(val_loader)}")
print(f"Test-like val batches: {len(test_like_val_loader)}")
```

```python
# ============================================
# CELL 8: TRAINING FUNCTIONS
# ============================================

def train_epoch(model, loader, optimizer, criterion, scaler):
    model.train()
    total_loss = 0
    all_preds = []
    all_labels = []
    
    pbar = tqdm(loader, desc='Training')
    for batch_idx, (inputs, labels) in enumerate(pbar):
        inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
        
        optimizer.zero_grad()
        
        # Mixed precision training
        with torch.amp.autocast('cuda'):
            outputs = model(inputs)
            loss = criterion(outputs, labels)
        
        scaler.scale(loss).backward()
        
        # Gradient clipping
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), 5.0)
        
        scaler.step(optimizer)
        scaler.update()
        
        total_loss += loss.item()
        
        # Get predictions
        _, predicted = outputs.max(1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        
        # Update progress bar
        pbar.set_postfix({'loss': total_loss/(batch_idx+1)})
    
    epoch_loss = total_loss / len(loader)
    if len(all_preds) > 0:
        epoch_f1 = f1_score(all_labels, all_preds, average='macro')
    else:
        epoch_f1 = 0.0
    
    return epoch_loss, epoch_f1

def validate(model, loader):
    model.eval()
    all_preds = []
    all_labels = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Validation'):
            inputs, labels = inputs.to(DEVICE), labels.to(DEVICE)
            
            outputs = model(inputs)
            _, predicted = outputs.max(1)
            
            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    
    if len(all_preds) > 0:
        epoch_f1 = f1_score(all_labels, all_preds, average='macro')
    else:
        epoch_f1 = 0.0
    
    return epoch_f1
```

```python
# ============================================
# CELL 9: TRAIN THE MODEL (FIXED)
# ============================================

# Initialize model
model = ImprovedCRNN(num_classes=len(GENRES)).to(DEVICE)
print(f"Model created with {sum(p.numel() for p in model.parameters()):,} parameters")

# Loss, optimizer, scheduler
criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
optimizer = torch.optim.AdamW(model.parameters(), lr=3e-4, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(
    optimizer, T_0=10, T_mult=2, eta_min=1e-6
)
scaler = torch.amp.GradScaler('cuda')

# Training loop
EPOCHS = 25
best_val_f1 = 0
best_test_like_f1 = 0
patience = 5
patience_counter = 0

for epoch in range(EPOCHS):
    print(f"\n{'='*50}")
    print(f"Epoch {epoch+1}/{EPOCHS}")
    print('='*50)
    
    # Train
    try:
        train_loss, train_f1 = train_epoch(model, train_loader, optimizer, criterion, scaler)
    except Exception as e:
        print(f"Error during training: {e}")
        continue
    
    # Validate on both validation sets
    try:
        val_f1 = validate(model, val_loader)
    except Exception as e:
        print(f"Error during validation: {e}")
        val_f1 = 0.0
    
    try:
        test_like_f1 = validate(model, test_like_val_loader)
    except Exception as e:
        print(f"Error during test-like validation: {e}")
        test_like_f1 = 0.0
    
    # Update scheduler
    scheduler.step()
    current_lr = optimizer.param_groups[0]['lr']
    
    print(f"\nTrain Loss: {train_loss:.4f}")
    print(f"Train F1: {train_f1:.4f}")
    print(f"Val F1 (standard): {val_f1:.4f}")
    print(f"Val F1 (test-like): {test_like_f1:.4f}")
    print(f"Learning Rate: {current_lr:.6f}")
    
    # Save based on test-like validation F1
    if test_like_f1 > best_test_like_f1:
        best_test_like_f1 = test_like_f1
        best_val_f1 = val_f1
        patience_counter = 0
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'test_like_f1': test_like_f1,
        }, 'best_model.pth')
        print(f"✓ Saved best model (test-like F1: {test_like_f1:.4f})")
    else:
        patience_counter += 1
    
    # Early stopping
    if patience_counter >= patience:
        print(f"\nEarly stopping triggered after {epoch+1} epochs")
        break

print(f"\n{'='*50}")
print(f"Best test-like validation F1: {best_test_like_f1:.4f}")
print(f"Best standard validation F1: {best_val_f1:.4f}")
print('='*50)
```

```python
# ============================================
# CELL 10: LOAD BEST MODEL AND PREDICT WITH TTA
# ============================================

# Load best model
if os.path.exists('best_model.pth'):
    checkpoint = torch.load('best_model.pth')
    model.load_state_dict(checkpoint['model_state_dict'])
    model.eval()
    print(f"Loaded best model with test-like validation F1: {checkpoint['test_like_f1']:.4f}")
else:
    print("No best model found, using current model")

# Load test data
test_df = pd.read_csv(TEST_CSV)
print(f"Test samples: {len(test_df)}")
print(f"Test columns: {test_df.columns.tolist()}")
print(f"First few rows:\n{test_df.head()}")

# Test Time Augmentation function
def predict_with_tta(model, audio_path, num_crops=5):
    """Make prediction with Test Time Augmentation"""
    try:
        # Load audio
        audio, _ = librosa.load(audio_path, sr=SR)
        
        logits_list = []
        
        for _ in range(num_crops):
            # Random crop
            if len(audio) > MAX_LEN:
                start = random.randint(0, len(audio) - MAX_LEN)
                crop = audio[start:start + MAX_LEN]
            else:
                crop = np.pad(audio, (0, MAX_LEN - len(audio)))
            
            # Extract features
            features = extract_melspectrogram(crop)
            features = torch.FloatTensor(features).unsqueeze(0).unsqueeze(0).to(DEVICE)
            
            # Forward pass
            with torch.no_grad():
                outputs = model(features)
                logits_list.append(outputs)
        
        # Average logits
        avg_logits = torch.mean(torch.cat(logits_list), dim=0)
        prediction = torch.argmax(avg_logits, dim=1).item()
        
        return prediction
    except Exception as e:
        print(f"Error predicting {audio_path}: {e}")
        return random.randint(0, 9)  # Fallback

# Generate predictions
print("\nGenerating predictions with TTA...")
predictions = []
TTA_CROPS = 5

for idx, row in tqdm(test_df.iterrows(), total=len(test_df), desc="Predicting"):
    # Handle different possible column names
    if 'filename' in test_df.columns:
        file_path = os.path.join(DATA_PATH, row['filename'])
    elif 'file' in test_df.columns:
        file_path = os.path.join(MASHUPS_PATH, row['file'])
    else:
        # Assume first column is id, second is filename
        file_col = test_df.columns[1]
        file_path = os.path.join(DATA_PATH, row[file_col])
    
    pred_idx = predict_with_tta(model, file_path, num_crops=TTA_CROPS)
    pred_genre = idx2label[pred_idx]
    
    predictions.append(pred_genre)
    
    # Debug first few
    if idx < 5:
        print(f"Sample {idx+1}: ID={row['id'] if 'id' in row else idx}, File={os.path.basename(file_path)}, Predicted={pred_genre}")
```
