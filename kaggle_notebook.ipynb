```python
import os
import glob
import random
import numpy as np
import pandas as pd
import librosa
import torch
import torch.nn as nn
import torchaudio
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from tqdm import tqdm
```

```python
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SR = 22050
DURATION = 10
SAMPLES = SR * DURATION
N_MELS = 128

BATCH_SIZE = 32
EPOCHS = 20
LR = 1e-3
SEED = 42

torch.manual_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)
```

```python
print(DEVICE)
```

```python
base_train_path = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/genres_stems"

data = []

genres = sorted(os.listdir(base_train_path))

for genre in genres:
    genre_path = os.path.join(base_train_path, genre)

    if not os.path.isdir(genre_path):
        continue

    for song_folder in os.listdir(genre_path):
        song_path = os.path.join(genre_path, song_folder)

        if os.path.isdir(song_path):
            data.append({
                "genre": genre,
                "song_folder": song_path
            })

train_df = pd.DataFrame(data)

print("Total training samples:", len(train_df))
```

```python
GENRES = sorted(train_df["genre"].unique())

label2idx = {g:i for i,g in enumerate(GENRES)}
idx2label = {i:g for g,i in label2idx.items()}

train_df["label"] = train_df["genre"].map(label2idx)
```

```python
mel_transform = torchaudio.transforms.MelSpectrogram(
    sample_rate=SR,
    n_fft=2048,
    hop_length=512,
    n_mels=N_MELS
)

db_transform = torchaudio.transforms.AmplitudeToDB()
```

```python
class AudioDataset(Dataset):
    def __init__(self, df, train=True):
        self.df = df.reset_index(drop=True)
        self.train = train
        
        # group by genre column
        self.genre_groups = self.df.groupby("genre")

        # collect ESC-50 noise files
        self.noise_files = glob.glob(
            "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/ESC-50-master/audio/*.wav"
        )

    def __len__(self):
        return len(self.df)

    def random_stem_mix(self, genre):
        songs = self.genre_groups.get_group(genre)["song_folder"].values
        
        stems = ["bass.wav", "drums.wav", "other.wav", "vocals.wav"]
        signals = []

        # Load all stems first
        for stem in stems:
            folder = random.choice(songs)
            path = os.path.join(folder, stem)
            signal, _ = librosa.load(path, sr=SR)
            signals.append(signal)

        # 🔥 Make all stems same length
        min_len = min(len(s) for s in signals)
        signals = [s[:min_len] for s in signals]

        # Mix safely
        mixture = np.sum(signals, axis=0)

        # Normalize
        mixture = mixture / (np.max(np.abs(mixture)) + 1e-6)

        return mixture

    def add_noise(self, signal):
        noise_path = random.choice(self.noise_files)
        noise, _ = librosa.load(noise_path, sr=SR)

        if len(noise) < len(signal):
            noise = np.pad(noise, (0, len(signal) - len(noise)))
        else:
            noise = noise[:len(signal)]

        noise_level = random.uniform(0.05, 0.15)
        return signal + noise_level * noise

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        genre = row["genre"]

        signal = self.random_stem_mix(genre)

        if self.train:
            signal = self.add_noise(signal)

        # crop/pad
        if len(signal) > SAMPLES:
            start = random.randint(0, len(signal) - SAMPLES)
            signal = signal[start:start+SAMPLES]
        else:
            signal = np.pad(signal, (0, SAMPLES - len(signal)))

        signal = torch.tensor(signal, dtype=torch.float32)

        mel = mel_transform(signal)
        mel = db_transform(mel)
        mel = (mel - mel.mean()) / (mel.std() + 1e-6)

        return mel.unsqueeze(0), row["label"]
```

```python
train_split, val_split = train_test_split(
    train_df,
    test_size=0.2,
    stratify=train_df["label"],
    random_state=SEED
)

train_ds = AudioDataset(train_split, train=True)
val_ds = AudioDataset(val_split, train=False)

# train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
# val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE)
train_loader = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE,
    num_workers=2,
    pin_memory=True
)
```

```python
class CNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()

        self.conv = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1,1))
        )

        self.dropout = nn.Dropout(0.4)
        self.fc = nn.Linear(128, num_classes)

    def forward(self, x):
        x = self.conv(x)
        x = x.view(x.size(0), -1)
        x = self.dropout(x)
        return self.fc(x)
```

```python
model = CNN(len(GENRES)).to(DEVICE)

criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
optimizer = torch.optim.Adam(model.parameters(), lr=LR)
```

```python
best_f1 = 0.0

for epoch in range(EPOCHS):

    model.train()
    total_loss = 0

    for x, y in tqdm(train_loader):
        x, y = x.to(DEVICE), y.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(x)
        loss = criterion(outputs, y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    # -------- Validation --------
    model.eval()
    preds, targets = [], []

    with torch.no_grad():
        for x, y in val_loader:
            x = x.to(DEVICE)
            outputs = model(x)

            p = torch.argmax(outputs, dim=1).cpu().numpy()

            preds.extend(p)
            targets.extend(y.numpy())

    f1 = f1_score(targets, preds, average="macro")

    print(f"Epoch {epoch+1} | Loss {total_loss/len(train_loader):.4f} | Val F1 {f1:.4f}")

    # -------- Save Best Model --------
    if f1 > best_f1:
        best_f1 = f1
        torch.save(model.state_dict(), "best_model.pth")
        print("Saving best model...")

print("Best Validation F1:", best_f1)
```

```python
import pandas as pd

test_df = pd.read_csv(
    "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/test.csv"
)

test_base = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup/mashups"

print(test_df.shape)
print(os.listdir(test_base)[:5])
```

```python
import os
import numpy as np
import torch
import soundfile as sf
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model.to(DEVICE)
model.eval()

# Move transforms to GPU
mel_transform = mel_transform.to(DEVICE)
db_transform = db_transform.to(DEVICE)

predictions = []

class TestDataset(Dataset):
    def __init__(self, df, base_path):
        self.df = df
        self.base_path = base_path

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        # row = self.df.iloc[idx]
        # path = os.path.join(self.base_path, row["filename"])

        # # Faster audio loading
        # signal, sr = sf.read(path)
        row = self.df.iloc[idx]

        filename = os.path.basename(row["filename"])
        path = os.path.join(self.base_path, filename)

        signal, sr = sf.read(path)

        # Convert to mono if stereo
        if len(signal.shape) > 1:
            signal = np.mean(signal, axis=1)

        # Pad or trim
        if len(signal) > SAMPLES:
            signal = signal[:SAMPLES]
        else:
            signal = np.pad(signal, (0, SAMPLES - len(signal)))

        signal = torch.tensor(signal, dtype=torch.float32)

        return signal


test_dataset = TestDataset(test_df, test_base)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,          # Larger batch for faster inference
    shuffle=False,
    num_workers=4,          # Increase workers
    pin_memory=True
)

with torch.no_grad():
    for signals in tqdm(test_loader):

        signals = signals.to(DEVICE)

        # Compute mel on GPU
        mels = mel_transform(signals)
        mels = db_transform(mels)

        # Normalize
        mels = (mels - mels.mean(dim=(1,2), keepdim=True)) / (
            mels.std(dim=(1,2), keepdim=True) + 1e-6
        )

        mels = mels.unsqueeze(1)  # Add channel dim

        outputs = model(mels)

        preds = torch.argmax(outputs, dim=1).cpu().numpy()

        predictions.extend([idx2label[p] for p in preds])
```
