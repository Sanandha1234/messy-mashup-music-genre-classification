```python
# ============================================================================
# COMPLETE PIPELINE: DATA2VEC TRAINING + 3-MODEL ENSEMBLE
# ============================================================================

# ============================================================================
# IMPORTS
# ============================================================================
import os
import random
import numpy as np
import pandas as pd
from tqdm import tqdm
import glob
import warnings
import datetime
from collections import defaultdict

# Audio processing
import librosa
import torchaudio
import torchaudio.transforms as T

# PyTorch
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Sklearn
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, accuracy_score

# Transformers
from transformers import (
    HubertModel, 
    AutoFeatureExtractor, 
    ASTForAudioClassification,
    Data2VecAudioModel,
    Data2VecAudioForSequenceClassification,
)

# Visualization
import matplotlib.pyplot as plt

# Wandb
import wandb

# Suppress warnings
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

# Set seed
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

# Device
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {DEVICE}")

# Audio constants
SR = 22050
SR_TRANSFORMER = 16000
DURATION = 30
MAX_LEN = SR * DURATION
MAX_LEN_TRANSFORMER = SR_TRANSFORMER * DURATION

# Training constants
BATCH_SIZE_HUBERT = 4
BATCH_SIZE_AST = 3
BATCH_SIZE_DATA2VEC = 3

# Genres
GENRES = ["blues", "classical", "country", "disco", "hiphop",
          "jazz", "metal", "pop", "reggae", "rock"]
label2idx = {g: i for i, g in enumerate(GENRES)}
idx2label = {i: g for g, i in label2idx.items()}
NUM_CLASSES = len(GENRES)

# Paths
DATA_PATH = "/kaggle/input/jan-2026-dl-gen-ai-project/messy_mashup"
STEMS_PATH = os.path.join(DATA_PATH, "genres_stems")
NOISE_PATH = os.path.join(DATA_PATH, "ESC-50-master", "audio")
TEST_CSV = os.path.join(DATA_PATH, "test.csv")

print(f"Genres: {GENRES}")
print(f"Num classes: {NUM_CLASSES}")

# ============================================================================
# AUDIO PROCESSING FUNCTIONS
# ============================================================================

def load_audio(file_path, target_sr=SR):
    """Load audio file."""
    try:
        audio, sr = librosa.load(file_path, sr=target_sr, mono=True)
        return audio
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return np.zeros(target_sr * 30)

def add_esc50_noise(audio, noise_files, num_noises=(1, 2), snr_range=(8, 25)):
    """Add ESC-50 noise to audio."""
    if len(noise_files) == 0 or random.random() > 0.5:
        return audio
    
    if isinstance(audio, torch.Tensor):
        audio = audio.numpy()
    
    num_to_add = random.randint(*num_noises)
    noisy_audio = audio.copy()
    audio_len = len(audio)
    
    for _ in range(num_to_add):
        noise_path = random.choice(noise_files)
        noise, _ = librosa.load(noise_path, sr=SR)
        
        if len(noise) < audio_len:
            start_pos = random.randint(0, audio_len - len(noise))
            noise_padded = np.zeros(audio_len)
            noise_padded[start_pos:start_pos + len(noise)] = noise
        else:
            start_pos = random.randint(0, len(noise) - audio_len)
            noise_padded = noise[start_pos:start_pos + audio_len]
        
        snr_db = random.uniform(*snr_range)
        signal_power = np.mean(audio ** 2)
        noise_power = np.mean(noise_padded ** 2)
        
        if noise_power > 0:
            scale = np.sqrt(signal_power / (noise_power * (10 ** (snr_db / 10))))
            noisy_audio += scale * noise_padded
    
    max_val = np.max(np.abs(noisy_audio))
    if max_val > 1.0:
        noisy_audio = noisy_audio / max_val * 0.95
    
    return noisy_audio

# ============================================================================
# LOAD DATA AND BUILD SAMPLES
# ============================================================================

# Load noise files
noise_files = glob.glob(os.path.join(NOISE_PATH, "*.wav"))
print(f"Found {len(noise_files)} noise files")

def build_training_samples(samples_per_genre=200):
    """Build training samples."""
    samples = []
    for genre in GENRES:
        genre_path = os.path.join(STEMS_PATH, genre)
        if not os.path.exists(genre_path):
            continue
        songs = os.listdir(genre_path)
        print(f"{genre}: {len(songs)} songs")
        
        for _ in range(samples_per_genre):
            if len(songs) < 4:
                selected = random.choices(songs, k=4)
            else:
                selected = random.sample(songs, 4)
            
            stems = []
            stem_names = ["drums.wav", "bass.wav", "vocals.wav", "other.wav"]
            for i, stem_name in enumerate(stem_names):
                stem_path = os.path.join(genre_path, selected[i], stem_name)
                stems.append(stem_path)
            
            samples.append({
                'stems': stems,
                'genre': genre,
                'label': label2idx[genre]
            })
    
    random.shuffle(samples)
    print(f"Total samples built: {len(samples)}")
    return samples

# Build samples
all_samples = build_training_samples(samples_per_genre=200)

# Split into train and validation
train_samples, val_samples = train_test_split(
    all_samples, test_size=0.15, 
    stratify=[s['label'] for s in all_samples], random_state=42
)

print(f"\nTrain samples: {len(train_samples)}")
print(f"Validation samples: {len(val_samples)}")

# ============================================================================
# DATASET CLASS
# ============================================================================

class TransformerDataset(Dataset):
    """Dataset for transformer models."""
    
    def __init__(self, samples, noise_files=None, train=True):
        self.samples = samples
        self.noise_files = noise_files or []
        self.train = train
        self.sr = SR_TRANSFORMER
        self.max_len = SR_TRANSFORMER * DURATION
        
        # Group by label for cross-song mixing
        self.label_to_samples = defaultdict(list)
        for sample in samples:
            self.label_to_samples[sample['label']].append(sample['stems'])
    
    def __len__(self):
        return len(self.samples)
    
    def load_audio(self, path):
        """Load and resample audio to 16kHz."""
        audio = load_audio(path, target_sr=SR)
        audio_tensor = torch.tensor(audio).float()
        audio_tensor = torchaudio.functional.resample(audio_tensor, SR, self.sr)
        return audio_tensor
    
    def __getitem__(self, idx):
        sample = self.samples[idx]
        label = sample['label']
        
        # Training mode: Cross-song mixing
        if self.train:
            candidates = self.label_to_samples[label]
            if len(candidates) >= 4:
                selected = random.sample(candidates, 4) if len(candidates) >= 4 else random.choices(candidates, k=4)
                
                signals = []
                stem_names = ["drums.wav", "bass.wav", "vocals.wav", "other.wav"]
                
                for i, stem_name in enumerate(stem_names):
                    stem_folder = os.path.dirname(selected[i][0])
                    stem_path = os.path.join(stem_folder, stem_name)
                    sig = self.load_audio(stem_path)
                    gain = random.uniform(0.6, 1.5)
                    signals.append(sig * gain)
            else:
                signals = []
                for stem_path in sample['stems']:
                    sig = self.load_audio(stem_path)
                    gain = random.uniform(0.6, 1.5)
                    signals.append(sig * gain)
        
        # Validation mode: No augmentation
        else:
            signals = []
            for stem_path in sample['stems']:
                sig = self.load_audio(stem_path)
                signals.append(sig)
        
        # Align to shortest stem
        min_len = min(len(s) for s in signals)
        signals = [s[:min_len] for s in signals]
        
        # Mix stems
        mixed = torch.stack(signals).mean(dim=0)
        
        # Normalize
        mixed = mixed / (mixed.abs().max() + 1e-6)
        
        # Add noise during training
        if self.train and len(self.noise_files) > 0 and random.random() > 0.3:
            mixed_np = add_esc50_noise(mixed.numpy(), self.noise_files)
            mixed = torch.tensor(mixed_np).float()
            mixed = mixed / (mixed.abs().max() + 1e-6)
        
        # Pad or trim to fixed length
        if len(mixed) > self.max_len:
            if self.train:
                start = random.randint(0, len(mixed) - self.max_len)
                mixed = mixed[start:start + self.max_len]
            else:
                start = (len(mixed) - self.max_len) // 2
                mixed = mixed[start:start + self.max_len]
        else:
            mixed = F.pad(mixed, (0, self.max_len - len(mixed)))
        
        return mixed, label

# Create datasets
train_ds = TransformerDataset(train_samples, noise_files=noise_files, train=True)
val_ds = TransformerDataset(val_samples, noise_files=[], train=False)

print(f"\n Dataset created")
print(f"  Train samples: {len(train_ds)}")
print(f"  Val samples: {len(val_ds)}")

# ============================================================================
# WANDB LOGIN
# ============================================================================

# Your WandB API key
WANDB_API_KEY = "REDACTED_WANDB_KEY"

# Set environment variable and login
os.environ["WANDB_API_KEY"] = WANDB_API_KEY

try:
    wandb.login()
    print(" Wandb logged in successfully!")
    USE_WANDB = True
except:
    print(" Could not login to wandb, training without logging")
    USE_WANDB = False

# ============================================================================
# DATA2VEC TRAINING (20 EPOCHS)
# ============================================================================

print("\n" + "="*80)
print(" DATA2VEC TRAINING (20 EPOCHS)")

class FixedData2VecClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        from transformers import Data2VecAudioModel
        self.data2vec = Data2VecAudioModel.from_pretrained("facebook/data2vec-audio-base-960h")
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(768, NUM_CLASSES)
        
        # Freeze first 2 layers
        layers_to_freeze = self.data2vec.encoder.layers[:2]
        for layer in layers_to_freeze:
            for param in layer.parameters():
                param.requires_grad = False
        
        print(f" Frozen first 2 layers of Data2Vec encoder")
    
    def forward(self, x):
        outputs = self.data2vec(x)
        features = outputs.last_hidden_state.mean(dim=1)
        return self.classifier(self.dropout(features))

# Initialize model
data2vec_model = FixedData2VecClassifier().to(DEVICE)
print(f"Model parameters: {sum(p.numel() for p in data2vec_model.parameters()):,}")

# Optimizer and loss
optimizer = torch.optim.AdamW(data2vec_model.parameters(), lr=3e-5, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss(label_smoothing=0.05)
scaler = torch.cuda.amp.GradScaler()

# Dataloaders
train_loader_data2vec = DataLoader(
    train_ds,
    batch_size=BATCH_SIZE_AST,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader_data2vec = DataLoader(
    val_ds,
    batch_size=BATCH_SIZE_AST,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print(f" Dataloaders created")
print(f"  Train batches: {len(train_loader_data2vec)}")
print(f"  Val batches: {len(val_loader_data2vec)}")

# Initialize wandb
if USE_WANDB:
    wandb.init(
        project="DL-24f2008134-notebook-t12026",
        name=f"data2vec-20-epochs-{datetime.datetime.now().strftime('%Y%m%d-%H%M')}",
        tags=["data2vec", "20-epochs", "final"],
        config={
            "model_type": "Data2Vec Audio",
            "frozen_layers": 2,
            "learning_rate": 3e-5,
            "batch_size": BATCH_SIZE_AST,
            "epochs": 20,
            "optimizer": "AdamW",
            "label_smoothing": 0.05
        }
    )
    wandb.watch(data2vec_model, log="all", log_freq=10)

# Training loop
best_f1 = 0
training_losses = []
val_f1_scores = []

for epoch in range(20):
    print(f"\n📅 Epoch {epoch+1}/20")
    
    # Training
    data2vec_model.train()
    train_loss = 0
    train_preds, train_labels = [], []
    
    for x, y in tqdm(train_loader_data2vec, desc='Training'):
        x, y = x.to(DEVICE), y.to(DEVICE)
        optimizer.zero_grad()
        
        with torch.cuda.amp.autocast():
            out = data2vec_model(x)
            loss = criterion(out, y)
        
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        
        train_loss += loss.item()
        train_preds.extend(out.argmax(1).detach().cpu().numpy())
        train_labels.extend(y.cpu().numpy())
    
    train_f1 = f1_score(train_labels, train_preds, average='macro')
    avg_train_loss = train_loss / len(train_loader_data2vec)
    training_losses.append(avg_train_loss)
    
    # Validation
    data2vec_model.eval()
    val_preds, val_labels = [], []
    with torch.no_grad():
        for x, y in tqdm(val_loader_data2vec, desc='Validation'):
            x, y = x.to(DEVICE), y.to(DEVICE)
            out = data2vec_model(x)
            val_preds.extend(out.argmax(1).cpu().numpy())
            val_labels.extend(y.cpu().numpy())
    
    val_f1 = f1_score(val_labels, val_preds, average='macro')
    val_f1_scores.append(val_f1)
    
    # Log to wandb
    if USE_WANDB:
        wandb.log({
            "epoch": epoch + 1,
            "train_loss": avg_train_loss,
            "train_f1": train_f1,
            "val_f1": val_f1,
            "learning_rate": optimizer.param_groups[0]['lr']
        })
    
    print(f"  Train Loss: {avg_train_loss:.4f}, Train F1: {train_f1:.4f}")
    print(f"  Val F1: {val_f1:.4f}")
    
    if val_f1 > best_f1:
        best_f1 = val_f1
        torch.save(data2vec_model.state_dict(), 'data2vec_best.pth')
        if USE_WANDB:
            wandb.run.summary["best_val_f1"] = best_f1
        print(f" Saved best model (F1: {best_f1:.4f})")
    
    # Early stopping if plateauing after epoch 12
    if epoch > 12:
        if len(val_f1_scores) >= 4 and val_f1_scores[-1] <= max(val_f1_scores[-4:-1]) * 0.98:
            print(f"\n Early stopping - F1 plateauing at {val_f1:.4f}")
            break

if USE_WANDB:
    wandb.finish()

print("\n" + "="*80)
print(f" Data2Vec Training Complete!")
print(f"  Best Validation F1: {best_f1:.4f}")
print("="*80)

# Save training history
np.save('data2vec_training_losses.npy', training_losses)
np.save('data2vec_val_f1_scores.npy', val_f1_scores)

# ============================================================================
# LOAD ALL MODELS
# ============================================================================
print(" LOADING ALL MODELS")
print("="*80)

test_df = pd.read_csv(TEST_CSV)
print(f"Test samples: {len(test_df)}")

# Feature extractor for AST
feature_extractor = AutoFeatureExtractor.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593")

# Define HuBERT Model
class HubertClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.hubert = HubertModel.from_pretrained("facebook/hubert-base-ls960")
        self.dropout = nn.Dropout(0.3)
        self.classifier = nn.Linear(768, NUM_CLASSES)
        for param in self.hubert.encoder.layers[:2].parameters():
            param.requires_grad = False
    
    def forward(self, x):
        out = self.hubert(x).last_hidden_state
        pooled = out.mean(dim=1)
        return self.classifier(self.dropout(pooled))

# Load AST model
print("\n Loading AST model")
ast_model = ASTForAudioClassification.from_pretrained(
    "MIT/ast-finetuned-audioset-10-10-0.4593",
    num_labels=NUM_CLASSES,
    ignore_mismatched_sizes=True
).to(DEVICE)
if os.path.exists('ast_best.pth'):
    ast_model.load_state_dict(torch.load('ast_best.pth', map_location=DEVICE))
    print(" AST model loaded")
else:
    print(" AST model not found!")

ast_model.eval()

# Load HuBERT model
print("\n Loading HuBERT model")
hubert_model = HubertClassifier().to(DEVICE)
if os.path.exists('hubert_best.pth'):
    hubert_model.load_state_dict(torch.load('hubert_best.pth', map_location=DEVICE))
    print(" HuBERT model loaded")
else:
    print(" HuBERT model not found!")

hubert_model.eval()

# Load Data2Vec model
print("\n Loading Data2Vec model")
data2vec_model = FixedData2VecClassifier().to(DEVICE)
if os.path.exists('data2vec_best.pth'):
    data2vec_model.load_state_dict(torch.load('data2vec_best.pth', map_location=DEVICE))
    print(f" Data2Vec model loaded (Best F1: {best_f1:.4f})")
else:
    print(" Data2Vec model not found!")

data2vec_model.eval()

# ============================================================================
# GENERATE PREDICTIONS
# ============================================================================

print("\n" + "="*80)
print(" GENERATING PREDICTIONS")

def process_audio_for_ast(audio_path):
    """Process audio for AST model"""
    audio, _ = librosa.load(audio_path, sr=SR)
    
    if len(audio) > MAX_LEN:
        start = (len(audio) - MAX_LEN) // 2
        audio = audio[start:start + MAX_LEN]
    else:
        audio = np.pad(audio, (0, MAX_LEN - len(audio)))
    
    audio_tensor = torch.tensor(audio).float()
    audio_tensor = torchaudio.functional.resample(audio_tensor, SR, SR_TRANSFORMER)
    audio_np = audio_tensor.numpy()
    
    inputs = feature_extractor(
        audio_np, 
        sampling_rate=SR_TRANSFORMER, 
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=SR_TRANSFORMER * DURATION
    )
    return inputs["input_values"]

def process_audio_for_others(audio_path):
    """Process audio for HuBERT and Data2Vec"""
    audio, _ = librosa.load(audio_path, sr=SR)
    
    if len(audio) > MAX_LEN:
        start = (len(audio) - MAX_LEN) // 2
        audio = audio[start:start + MAX_LEN]
    else:
        audio = np.pad(audio, (0, MAX_LEN - len(audio)))
    
    audio_tensor = torch.tensor(audio).float()
    audio_tensor = torchaudio.functional.resample(audio_tensor, SR, SR_TRANSFORMER)
    return audio_tensor.unsqueeze(0)

print("\n Generating predictions:-")
ast_probs = []
hubert_probs = []
data2vec_probs = []

with torch.no_grad():
    for idx, row in tqdm(test_df.iterrows(), total=len(test_df), desc="Processing"):
        audio_path = os.path.join(DATA_PATH, row['filename'])
        
        # AST prediction
        input_values = process_audio_for_ast(audio_path).to(DEVICE)
        ast_out = ast_model(input_values).logits
        ast_probs.append(torch.softmax(ast_out, dim=1).cpu().numpy()[0])
        
        # HuBERT and Data2Vec prediction
        x = process_audio_for_others(audio_path).to(DEVICE)
        
        hubert_out = hubert_model(x)
        if hasattr(hubert_out, 'logits'):
            hubert_out = hubert_out.logits
        hubert_probs.append(torch.softmax(hubert_out, dim=1).cpu().numpy()[0])
        
        data2vec_out = data2vec_model(x)
        data2vec_probs.append(torch.softmax(data2vec_out, dim=1).cpu().numpy()[0])

# Convert to numpy arrays
ast_probs = np.array(ast_probs)
hubert_probs = np.array(hubert_probs)
data2vec_probs = np.array(data2vec_probs)

print(f"\n Predictions are:-")
print(f"  AST: {ast_probs.shape}")
print(f"  HuBERT: {hubert_probs.shape}")
print(f"  Data2Vec: {data2vec_probs.shape}")

# Save predictions
np.save('ast_probs.npy', ast_probs)
np.save('hubert_probs.npy', hubert_probs)
np.save('data2vec_probs.npy', data2vec_probs)
print(" Predictions saved")

# ============================================================================
# CREATE ENSEMBLE SUBMISSION
# ============================================================================
print(" CREATING ENSEMBLE SUBMISSION")
print("="*80)

# Set ensemble weights based on Data2Vec performance
if best_f1 > 0.85:
    weights = {'ast': 0.55, 'hubert': 0.15, 'data2vec': 0.30}
    print(" Using high Data2Vec weights (F1 > 0.85)")
elif best_f1 > 0.75:
    weights = {'ast': 0.60, 'hubert': 0.15, 'data2vec': 0.25}
    print(" Using medium-high Data2Vec weights (F1 > 0.75)")
elif best_f1 > 0.65:
    weights = {'ast': 0.65, 'hubert': 0.15, 'data2vec': 0.20}
    print(" Using medium Data2Vec weights (F1 > 0.65)")
elif best_f1 > 0.50:
    weights = {'ast': 0.70, 'hubert': 0.15, 'data2vec': 0.15}
    print(" Using low Data2Vec weights (F1 > 0.50)")
else:
    weights = {'ast': 0.70, 'hubert': 0.30, 'data2vec': 0.00}
    print(" Using AST+HuBERT only (Data2Vec F1 too low)")

print(f"\n Final Ensemble Weights:")
print(f"  AST: {weights['ast']:.2f}")
print(f"  HuBERT: {weights['hubert']:.2f}")
print(f"  Data2Vec: {weights['data2vec']:.2f}")

# Create ensemble
ensemble_probs = (weights['ast'] * ast_probs + 
                  weights['hubert'] * hubert_probs + 
                  weights['data2vec'] * data2vec_probs)

final_preds = ensemble_probs.argmax(1)
final_genres = [idx2label[p] for p in final_preds]

# Create submission
submission = pd.DataFrame({
    'id': test_df['id'],
    'genre': final_genres
})

# Save submission
submission_filename = f'submission_3model_ensemble_f1{best_f1:.3f}.csv'
submission.to_csv(submission_filename, index=False)

print(f"\n Submission saved as '{submission_filename}'")
print(f"   Shape: {submission.shape}")

# Display statistics
print("\n Genre distribution in submission:")
genre_counts = submission['genre'].value_counts().sort_index()
for genre, count in genre_counts.items():
    percentage = (count / len(submission)) * 100
    print(f"  {genre}: {count} ({percentage:.1f}%)")
print(submission.head(10))

# Display confidence scores
print("\n Sample confidence scores:")
sample_indices = [0, 1, 2, 3, 4]
for idx in sample_indices:
    pred_genre = final_genres[idx]
    confidence = ensemble_probs[idx][final_preds[idx]]
    print(f"  Sample {idx}: {pred_genre} (confidence: {confidence:.4f})")

# ============================================================================
# FINAL SUMMARY
# ============================================================================

print(f"\n Model Performance Summary:")
print(f"  AST Validation F1: 0.9769")
print(f"  AST Test Score: 0.815 ")
print(f"  HuBERT Validation F1: 0.88")
print(f"  HuBERT Test Score: 0.72")
print(f"  Data2Vec Validation F1: {best_f1:.4f}")
print(f"  Data2Vec Test Score: ~{best_f1 - 0.16:.4f} (estimated)")

print(f"\n Final Ensemble:")
print(f"  Weights: AST={weights['ast']:.2f}, HuBERT={weights['hubert']:.2f}, Data2Vec={weights['data2vec']:.2f}")
print(f"  Submission: {submission_filename}")
```
